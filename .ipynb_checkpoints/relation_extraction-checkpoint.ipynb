# Relation Extraction Pipeline

Extracts **typed entity→entity edges** from cached chunks. These edges form the
relation layer that PPR traverses — they are NOT reified as Fact nodes and are
NOT stored on `HAS_FACT` edges.

**Why this is a separate notebook from `fact_extraction.ipynb`:**
- PPR walks `RELATED_TO`-family edges between entity nodes.
- `HAS_FACT` is invisible to PPR by design — it hangs scalar attribute facts off entities.
- Relations (usesTool, worksFor, hasPart, …) have a second entity on the far end,
  so they ARE edges, not fact nodes. Scalar facts have no far entity, so they need nodes.

**Pipeline (3 stages — no normalizer needed):**
1. **Stage 1** — LLM extracts entity→entity triples; explicit instruction to skip scalars.
2. **Stage 2** — Object resolution: `raw_object_str` → canonical entity ID via `mention_to_canonical`.
3. **Stage 3** — Span re-verification + confidence scoring.

**Inputs:** `entity_cache.json`, `resolved_entities.json`  
**Output:** `relations_cache.json` — staged RelationEdge records ready for the graph layer.

> **Bi-temporal note:** Default = plain typed edge (current state). If a relation needs
> history, add `valid_from`/`valid_to` on the edge. Do NOT reify relation statements
> as Fact nodes unless you need a full audit trail AND still project a current edge for PPR.

## Setup & imports

````python
import json
import re
import sys
import hashlib
from pathlib import Path
from typing import Literal, NamedTuple

from dotenv import load_dotenv
from openai import OpenAI
from pydantic import BaseModel, Field

load_dotenv()
client = OpenAI()

ENTITY_CACHE    = Path("entity_cache.json")
RESOLVED_CACHE  = Path("resolved_entities.json")
RELATIONS_CACHE = Path("relations_cache.json")

EXTRACTOR_VERSION = "relation-extraction-v0/gpt-4.1-mini"

for p in (ENTITY_CACHE, RESOLVED_CACHE):
    if not p.exists():
        sys.exit(f"missing cache: {p}")
print("caches found ✓")
````

## RelationEdge contract

Both ends must resolve to canonical entity IDs. An unresolved object means the
target entity was not extracted in the entity step — logged as a recall miss.

````python
FACT_TYPE_VALUES = [
    "assertion",
    "definition",
    "opinion",
    "instruction",
    "metaphor",
    "prediction",
    "other_non_fact",
]


class RelationEdge(BaseModel):
    """Typed entity-to-entity edge. Keyed by canonical IDs on both ends.

    These are PPR-walkable edges in the relation layer.
    They are NOT reified as HAS_FACT attribute nodes.
    """

    relation_id:        str   # stable hash of (doc_id, char_start, relation_type, subject, object_raw)

    # endpoints — both must be canonical
    subject_entity_id:  str   # canonical, from the entity list
    relation_type:      str   # usesTool | worksFor | hasPart | …
    relation_type_raw:  str   = ""   # when relation_type="other"
    object_raw:         str   # verbatim object name from source
    object_entity_id:   str | None = None   # canonical, resolved via mention_to_canonical
    object_resolved:    bool  = False

    # fact classification (same gate as fact pipeline)
    fact_type:          str   = "assertion"

    # bi-temporal validity — opt-in per relation type; default = current state only
    valid_from:         str | None = None
    valid_to:           str | None = None
    temporal_inferred:  bool  = False

    # provenance
    doc_id:             str
    char_start:         int
    char_end:           int
    source_quote:       str

    # pipeline stamps
    span_verified:      bool  = False
    confidence:         float = 1.0
    extractor_version:  str   = EXTRACTOR_VERSION


class ChunkResult(NamedTuple):
    edges:             list[RelationEdge]
    non_facts:         list[dict]   # rejected by fact-type gate
    rejected_subjects: list[dict]   # subject not in canonical entity list
    unresolved_objects:list[dict]   # object not in mention_to_canonical


def make_relation_id(
    doc_id: str, char_start: int, relation_type: str, subject: str, object_raw: str
) -> str:
    key = f"{doc_id}|{char_start}|{relation_type}|{subject}|{object_raw}"
    return hashlib.sha1(key.encode()).hexdigest()[:16]


print("RelationEdge contract loaded ✓")
````

## Load caches

````python
resolved_data        = json.loads(RESOLVED_CACHE.read_text())
mention_to_canonical: dict[str, str] = resolved_data["mention_to_canonical"]
canonical_set = set(mention_to_canonical.values())
print(f"resolved entities : {resolved_data['total_entities']:>5}")
print(f"surface forms     : {len(mention_to_canonical):>5}")

entity_cache = json.loads(ENTITY_CACHE.read_text())
mentions     = entity_cache["mentions"]

_doc_cache: dict[str, str] = {}

def _doc_text(source_file: str) -> str:
    if source_file not in _doc_cache:
        p = Path(source_file)
        _doc_cache[source_file] = p.read_text(encoding="utf-8", errors="ignore") if p.exists() else ""
    return _doc_cache[source_file]


ChunkKey = tuple[str, str]
chunks: dict[ChunkKey, dict] = {}

for m in mentions:
    key: ChunkKey = (m["source_file"], str(m["episode_idx"]))
    if key in chunks:
        continue
    episode = m["episode"]
    doc     = _doc_text(m["source_file"])
    offset  = doc.find(episode)
    chunks[key] = {
        "doc_id":      m["source_file"],
        "episode_idx": str(m["episode_idx"]),
        "text":        episode,
        "char_start":  offset,
        "char_end":    offset + len(episode) if offset >= 0 else -1,
    }

chunk_entities: dict[ChunkKey, list[str]] = {k: [] for k in chunks}
for m in mentions:
    key = (m["source_file"], str(m["episode_idx"]))
    canon = mention_to_canonical.get(m["text"])
    if canon and canon not in chunk_entities[key]:
        chunk_entities[key].append(canon)

bad = sum(1 for c in chunks.values() if c["char_start"] < 0)
print(f"\nchunks            : {len(chunks):>5}")
print(f"bad char offsets  : {bad:>5}")
````

## Stage 1 — LLM relation extraction

The model receives the chunk text + known canonical entities.
It emits entity→entity triples only — explicit negative examples for scalars.
Both subject and object come from the text verbatim; object is resolved post-hoc.

````python
from dataclasses import dataclass


@dataclass
class RelationSet:
    """Holds one complete relation-type vocabulary for Stage 1."""
    name:    str
    types:   list[str]   # closed enum values
    glosses: str


FEW_SHOT_RELATION = """\
EXAMPLES:

KEEP (entity-to-entity relation):
  'team X uses SharePoint'          -> usesTool(team X, SharePoint)
  'GraphRAG has component Y'        -> hasPart(GraphRAG, Y)
  'Alice works for Acme Corp'       -> worksFor(Alice, Acme Corp)
  'Community detection is part of GraphRAG' -> partOf(Community detection, GraphRAG)

SKIP (scalar — handled in fact_extraction.ipynb):
  'the team has 12 people'          -> headcount = 12, NOT a relation
  'founded in 2019'                 -> foundingDate, NOT a relation

SKIP (non-fact):
  'gold mining like exploding dynamite'  -> metaphor
  'conversations are bad by default'     -> opinion
"""

EXTRACT_MODEL = "gpt-4.1-mini"


def _build_system(rel_set: RelationSet) -> str:
    return f"""You extract TYPED ENTITY-TO-ENTITY RELATIONS from a text chunk.

A relation links two named entities: (subject) -[relation_type]-> (object).

You do NOT extract scalar attribute facts here.
  'team X has 12 people' is a scalar (headcount) — skip it.
  'team X uses SharePoint' is a relation (usesTool) — keep it.

{FEW_SHOT_RELATION}

FACT CLASSIFICATION — label every claim:
  assertion     — a checkable entity-to-entity relation (KEEP)
  definition    — explains what X is (SKIP)
  opinion       — normative claim (SKIP)
  instruction   — imperative (SKIP)
  metaphor      — analogy (SKIP)
  prediction    — speculative (SKIP)
  other_non_fact — not an assertion (SKIP)

You will receive:
- TEXT: the source chunk
- ENTITIES: the ONLY valid subjects (canonical entity IDs)

For each relation:
- subject_entity_id: copy VERBATIM from the ENTITIES list
- relation_type: from the closed enum; use 'other' + fill relation_type_raw if none fits
- raw_object_str: the object entity name VERBATIM from the text (will be resolved post-hoc)
- valid_from / valid_to: ISO dates ONLY if explicitly stated; else null
- temporal_inferred: true if you inferred timing rather than reading it

Relation type glosses:
{rel_set.glosses}"""


def _build_schema(rel_set: RelationSet) -> dict:
    return {
        "type": "object",
        "additionalProperties": False,
        "properties": {
            "relations": {
                "type": "array",
                "items": {
                    "type": "object",
                    "additionalProperties": False,
                    "properties": {
                        "subject_entity_id": {"type": "string"},
                        "relation_type":     {"type": "string", "enum": rel_set.types},
                        "relation_type_raw": {"type": "string"},
                        "raw_object_str":    {"type": "string"},
                        "fact_type":         {"type": "string", "enum": FACT_TYPE_VALUES},
                        "valid_from":        {"type": ["string", "null"]},
                        "valid_to":          {"type": ["string", "null"]},
                        "temporal_inferred": {"type": "boolean"},
                    },
                    "required": [
                        "subject_entity_id", "relation_type", "relation_type_raw",
                        "raw_object_str", "fact_type",
                        "valid_from", "valid_to", "temporal_inferred",
                    ],
                },
            }
        },
        "required": ["relations"],
    }


def extract_relations_from_chunk(
    chunk: dict, entities: list[str], rel_set: RelationSet
) -> list[dict]:
    """Stage 1: LLM extracts entity-to-entity relation triples."""
    if not entities:
        return []
    user_msg = (
        f"TEXT:\n{chunk['text']}\n\n"
        "ENTITIES:\n" + "\n".join(f"- {e}" for e in entities)
    )
    resp = client.chat.completions.create(
        model=EXTRACT_MODEL,
        temperature=0,
        messages=[
            {"role": "system", "content": _build_system(rel_set)},
            {"role": "user",   "content": user_msg},
        ],
        response_format={
            "type": "json_schema",
            "json_schema": {
                "name": "relations",
                "strict": True,
                "schema": _build_schema(rel_set),
            },
        },
    )
    return json.loads(resp.choices[0].message.content)["relations"]


# ── v0 relation set ───────────────────────────────────────────────────────────
# From v0 predicate schema + Wikidata structural predicates.
# PIDs: hasPart P527 ✓, partOf P361 ✓, uses P2283 ✓, produces P1056 ✓,
#       instanceOf P31 ✓, subclassOf P279 ✓, foundedBy P112 ✓

_REL_TYPES_V0 = [
    # v0 predicates (now as edges)
    "usesTool",       # subject uses tool / system / method X
    "ownerOf",        # subject is accountable for X
    "worksFor",       # person works for org or team X
    "documentedIn",   # knowledge/process X lives in source Y
    "partOf",         # X is part of larger org or system
    "responsibleFor", # subject leads or performs project/task X
    # Wikidata structural (formerly in wikidata-v1 fact set — now correctly edges)
    "hasPart",        # P527 — X has component/part/step Y
    "instanceOf",     # P31  — X is an instance of type Y
    "subclassOf",     # P279 — X is a subclass of Y
    "uses",           # P2283 — X uses Y (synonym for usesTool, broader)
    "produces",       # P1056 — X generates/outputs Y
    "foundedBy",      # P112  — org X founded by Y
    "other",
]

_REL_GLOSSES_V0 = "\n".join([
    "usesTool       — subject (person/team/process) uses tool, system, or method X",
    "ownerOf        — subject is ACCOUNTABLE for X (can be blamed/credited for it)",
    "worksFor       — person X works for org or team Y",
    "documentedIn   — knowledge or process X is documented in source Y",
    "partOf         — X is part of larger org, system, or department Y",
    "responsibleFor — subject leads or executes project/task X",
    "hasPart        — X has component, step, or sub-system Y (P527)",
    "instanceOf     — X is an instance or example of type Y (P31)",
    "subclassOf     — X is a subclass or kind of Y, more specific (P279)",
    "uses           — X uses Y (broader than usesTool; any usage) (P2283)",
    "produces       — X generates, outputs, or produces Y (P1056)",
    "foundedBy      — org X was founded by person/entity Y (P112)",
    "other          — no relation type fits; fill relation_type_raw",
])

REL_SET_V0 = RelationSet(
    name="relation-v0",
    types=_REL_TYPES_V0,
    glosses=_REL_GLOSSES_V0,
)

print(f"Stage 1 ready ✓  ({len(_REL_TYPES_V0)} relation types)")
````

## Stage 2 — Object resolution

Resolves `raw_object_str` → canonical entity ID via `mention_to_canonical`.
Unresolved objects are kept in the edge record but flagged — they mean the object
entity was not extracted in the upstream entity step (a recall miss, not an error here).

````python
def resolve_object(raw_object_str: str) -> tuple[str | None, bool]:
    """Stage 2: resolve raw object string to canonical entity ID.

    Returns (canonical_id, resolved). If not resolved, canonical_id is None.
    """
    canon = mention_to_canonical.get(raw_object_str)
    if canon:
        return canon, True
    # case-insensitive fallback
    raw_lower = raw_object_str.lower()
    for surface, cid in mention_to_canonical.items():
        if surface.lower() == raw_lower:
            return cid, True
    return None, False


print("Stage 2 object resolver ready ✓")
````

## Stage 3 — Span verification + confidence

Checks that both the subject name (or a fragment) and the raw object string appear
in `source_quote`. Relations are harder to ground than scalars — the span check
is a best-effort signal, not a hard gate.

**Confidence formula (not logprob-based):**
```
start                    1.00
object_unresolved       -0.20  object not in mention_to_canonical
span_verified=False     -0.30  neither subject nor object found verbatim in quote
temporal_inferred       -0.10  valid-time inferred, not stated
relation_type=other     -0.10  no named relation type fit
```

````python
_DIGITS_RE = re.compile(r"\d+")


def verify_relation_span(
    raw_object_str: str,
    source_quote: str,
    valid_from: str | None,
    valid_to: str | None,
) -> dict:
    """Stage 3: check raw_object_str appears in source_quote."""
    q_lower  = source_quote.lower()
    obj_lower = raw_object_str.lower().strip()
    span_verified = bool(obj_lower) and obj_lower in q_lower

    temporal_inferred = False
    for date_val in filter(None, [valid_from, valid_to]):
        digits = _DIGITS_RE.findall(date_val)
        if digits and not all(d in source_quote for d in digits):
            temporal_inferred = True
            break

    return {"span_verified": span_verified, "temporal_inferred": temporal_inferred}


def score_confidence(
    object_resolved: bool,
    span_verified: bool,
    temporal_inferred: bool,
    relation_type: str,
) -> float:
    """Stage 3 confidence for relation edges."""
    score = 1.00
    if not object_resolved:         score -= 0.20
    if not span_verified:           score -= 0.30
    if temporal_inferred:           score -= 0.10
    if relation_type == "other":    score -= 0.10
    return round(max(0.0, score), 2)


print("Stage 3 verify + confidence ready ✓")
print("  formula: 1.0 - 0.20(unresolved) - 0.30(span_fail) - 0.10(temporal) - 0.10(other)")
````

## Full pipeline — assemble RelationEdge per LLM extraction

````python
def process_chunk(
    chunk: dict,
    entities: list[str],
    rel_set: RelationSet = REL_SET_V0,
) -> ChunkResult:
    """Run all stages on one chunk and return RelationEdges."""
    raw_items = extract_relations_from_chunk(chunk, entities, rel_set)

    edges:              list[RelationEdge] = []
    non_facts:          list[dict]         = []
    rejected_subjects:  list[dict]         = []
    unresolved_objects: list[dict]         = []

    entity_set = set(entities)

    for rr in raw_items:
        fact_type   = rr.get("fact_type", "assertion")
        rel_type    = rr["relation_type"]
        raw_obj     = rr["raw_object_str"]

        # Gate: non-assertions into non_fact bucket
        if fact_type != "assertion":
            non_facts.append({
                "fact_type":         fact_type,
                "subject_entity_id": rr["subject_entity_id"],
                "relation_type":     rel_type,
                "raw_object_str":    raw_obj,
                "doc_id":            chunk["doc_id"],
            })
            continue

        # Fix 2: subject must be canonical
        subj = rr["subject_entity_id"]
        if subj not in entity_set:
            rejected_subjects.append({
                "subject_attempted": subj,
                "relation_type":     rel_type,
                "raw_object_str":    raw_obj,
                "doc_id":            chunk["doc_id"],
            })
            continue

        source_quote = chunk["text"]
        char_start   = chunk["char_start"]
        char_end     = chunk["char_end"]

        # Stage 2: object resolution
        obj_canon, obj_resolved = resolve_object(raw_obj)

        if not obj_resolved:
            unresolved_objects.append({
                "subject_entity_id": subj,
                "relation_type":     rel_type,
                "raw_object_str":    raw_obj,
                "doc_id":            chunk["doc_id"],
                "note":              "object not in mention_to_canonical — upstream recall miss",
            })
            # Still create the edge — just unflagged; caller decides what to do

        # Stage 3: verify + confidence
        verify = verify_relation_span(
            raw_obj, source_quote,
            rr.get("valid_from"), rr.get("valid_to")
        )
        temporal_inf = verify["temporal_inferred"] or rr.get("temporal_inferred", False)
        conf = score_confidence(
            object_resolved   = obj_resolved,
            span_verified     = verify["span_verified"],
            temporal_inferred = temporal_inf,
            relation_type     = rel_type,
        )

        edge = RelationEdge(
            relation_id       = make_relation_id(chunk["doc_id"], char_start, rel_type, subj, raw_obj),
            subject_entity_id = subj,
            relation_type     = rel_type,
            relation_type_raw = rr.get("relation_type_raw", ""),
            object_raw        = raw_obj,
            object_entity_id  = obj_canon,
            object_resolved   = obj_resolved,
            fact_type         = fact_type,
            valid_from        = rr.get("valid_from"),
            valid_to          = rr.get("valid_to"),
            temporal_inferred = temporal_inf,
            doc_id            = chunk["doc_id"],
            char_start        = char_start,
            char_end          = char_end,
            source_quote      = source_quote,
            span_verified     = verify["span_verified"],
            confidence        = conf,
        )
        edges.append(edge)

    return ChunkResult(
        edges=edges,
        non_facts=non_facts,
        rejected_subjects=rejected_subjects,
        unresolved_objects=unresolved_objects,
    )


print("pipeline assembled ✓")
````

## Demo run — 10 chunks

Shows relation edges extracted, resolution rate, and which Construction.md component
relations now land on `hasPart` / `produces` instead of being buried as Fact nodes.

````python
from collections import Counter

DEMO_CHUNKS = 10

demo_edges:     list[RelationEdge] = []
demo_non_facts: list[dict]         = []
demo_rejected:  list[dict]         = []
demo_unresolved:list[dict]         = []

chunk_items = [
    (key, chunk) for key, chunk in chunks.items()
    if chunk["char_start"] >= 0 and chunk_entities.get(key)
][:DEMO_CHUNKS]

for key, chunk in chunk_items:
    entities = chunk_entities[key]
    try:
        result = process_chunk(chunk, entities, REL_SET_V0)
        demo_edges.extend(result.edges)
        demo_non_facts.extend(result.non_facts)
        demo_rejected.extend(result.rejected_subjects)
        demo_unresolved.extend(result.unresolved_objects)
    except Exception as e:
        print(f"  chunk {key} failed: {e}")

total = len(demo_edges) + len(demo_non_facts) + len(demo_rejected)

print("=" * 60)
print(f"  RELATION EXTRACTION  ({DEMO_CHUNKS} chunks)")
print("=" * 60)
print(f"  total LLM claims    : {total}")
print(f"  relation edges kept : {len(demo_edges)}")
print(f"  non-facts rejected  : {len(demo_non_facts)}")
for ft, cnt in sorted(Counter(n['fact_type'] for n in demo_non_facts).items(), key=lambda x: -x[1]):
    print(f"    {ft:<20} {cnt}")
print(f"  invalid subjects    : {len(demo_rejected)}")
print()

# Resolution stats
resolved   = sum(1 for e in demo_edges if e.object_resolved)
unresolved = len(demo_unresolved)
print(f"  OBJECT RESOLUTION")
print(f"  resolved            : {resolved}/{len(demo_edges)} "
      f"({100*resolved//max(1,len(demo_edges))}%)")
print(f"  unresolved (logged) : {unresolved}")
if demo_unresolved[:3]:
    for u in demo_unresolved[:3]:
        print(f"    {u['subject_entity_id'][:25]:<25} -{u['relation_type']}-> {u['raw_object_str']!r}")
print()

# Confidence distribution
if demo_edges:
    confs = sorted(e.confidence for e in demo_edges)
    buckets = Counter(round(c, 1) for c in confs)
    print(f"  CONFIDENCE  min={confs[0]:.2f} max={confs[-1]:.2f} mean={sum(confs)/len(confs):.2f}")
    for score in sorted(buckets):
        print(f"    {score:.1f}  {'█' * buckets[score]} ({buckets[score]})")
    print()

# Construction.md — motivating case
const_edges = [e for e in demo_edges if "Construction" in e.doc_id]
if const_edges:
    print("=" * 60)
    print("  CONSTRUCTION.md edges (was 'ownerOf (has component)' before)")
    print("=" * 60)
    for e in const_edges:
        res = e.object_entity_id or "(unresolved)"
        print(f"  [{e.confidence:.2f}] {e.subject_entity_id[:22]:<22} "
              f"-{e.relation_type}-> {e.object_raw!r}")
        if e.object_entity_id:
            print(f"  {'':>7} object resolved: {e.object_entity_id}")
    print()

# All edges
print("=" * 60)
print(f"  ALL RELATION EDGES ({len(demo_edges)})")
print("=" * 60)
for e in sorted(demo_edges, key=lambda x: (-x.confidence, x.relation_type)):
    res_mark = "✓" if e.object_resolved else "?"
    sv_mark  = "✓" if e.span_verified   else "✗"
    print(f"  [{e.confidence:.2f}] res={res_mark} span={sv_mark}")
    print(f"    {e.subject_entity_id[:30]:<30} -{e.relation_type}-> {e.object_raw!r}")
    if e.object_entity_id:
        print(f"    object: {e.object_entity_id}")
    print(f"    source: {Path(e.doc_id).name}  [{e.char_start}:{e.char_end}]")
    print()
````

## Bulk run — all chunks → `relations_cache.json`

````python
from concurrent.futures import ThreadPoolExecutor, as_completed
from tqdm.notebook import tqdm as tqdm_nb

BULK_WORKERS = 8
SAVE_EVERY   = 50

eligible = [
    (key, chunk) for key, chunk in chunks.items()
    if chunk["char_start"] >= 0 and chunk_entities.get(key)
]
print(f"{len(eligible)} eligible chunks")

all_edges:     list[dict] = []
all_non_facts: list[dict] = []
all_unresolved:list[dict] = []

def _process(args: tuple) -> tuple[list[dict], list[dict], list[dict]]:
    key, chunk = args
    result = process_chunk(chunk, chunk_entities[key], REL_SET_V0)
    return (
        [e.model_dump(mode="json") for e in result.edges],
        result.non_facts,
        result.unresolved_objects,
    )

def _save() -> None:
    RELATIONS_CACHE.write_text(json.dumps(all_edges, ensure_ascii=False, indent=2))

with ThreadPoolExecutor(max_workers=BULK_WORKERS) as executor:
    futures = {executor.submit(_process, item): item for item in eligible}
    with tqdm_nb(total=len(eligible), desc="relation edges") as bar:
        for i, future in enumerate(as_completed(futures)):
            eb, nfb, urb = future.result()
            all_edges.extend(eb)
            all_non_facts.extend(nfb)
            all_unresolved.extend(urb)
            bar.update(1)
            if (i + 1) % SAVE_EVERY == 0:
                _save()

_save()

resolved_n    = sum(1 for e in all_edges if e.get("object_resolved"))
conf_vals     = [e["confidence"] for e in all_edges]
by_type: dict[str, int] = {}
for e in all_edges:
    by_type[e["relation_type"]] = by_type.get(e["relation_type"], 0) + 1
non_fact_types: dict[str, int] = {}
for n in all_non_facts:
    non_fact_types[n["fact_type"]] = non_fact_types.get(n["fact_type"], 0) + 1

print(f"\n{len(all_edges)} edges saved  +  {len(all_non_facts)} non-facts  +  {len(all_unresolved)} unresolved objects")
print(f"  object resolved : {resolved_n}/{len(all_edges)} ({100*resolved_n//max(1,len(all_edges))}%)")
print(f"  avg confidence  : {sum(conf_vals)/max(1,len(conf_vals)):.2f}")
print("\n  by relation type:")
for rt, cnt in sorted(by_type.items(), key=lambda x: -x[1]):
    print(f"    {rt:<18} {cnt}")
print("\n  non-fact types:")
for ft, cnt in sorted(non_fact_types.items(), key=lambda x: -x[1]):
    print(f"    {ft:<20} {cnt}")
````

## Graph write — next step

`relations_cache.json` is the staging area. To write to Neo4j:

```cypher
// Idempotent MERGE on relation_id
MATCH (s:Entity {canonical_id: $subject_entity_id})
MATCH (o:Entity {canonical_id: $object_entity_id})
MERGE (s)-[r:RELATED_TO {relation_id: $relation_id}]->(o)
SET r.relation_type  = $relation_type,
    r.source_quote   = $source_quote,
    r.doc_id         = $doc_id,
    r.confidence     = $confidence,
    r.valid_from     = $valid_from,
    r.valid_to       = $valid_to
```

Skip edges where `object_entity_id` is null (unresolved object — no target node to connect to).
