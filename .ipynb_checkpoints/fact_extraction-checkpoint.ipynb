# Typed Fact Extraction Pipeline

Turns cached chunks + resolved entities into **reified Fact records** ready for a bi-temporal graph.

**Pipeline:**
1. **Stage 1 — LLM extraction** — finds facts, emits `raw_value_str` verbatim. Never computes typed values.
2. **Stage 2 — Deterministic normalizer** — pure-Python parsing of raw strings into typed scalars. No LLM.
3. **Stage 3 — Span re-verification** — confirms the raw value actually appears in the source span.
4. **Stage 4 — Confidence scoring** — rule-based composite score from stages 1-3.

**Inputs (cached):**
- `entity_cache.json` — mention text, episode text, source_file
- `resolved_entities.json` — canonical IDs + `mention_to_canonical` lookup

**Output:** `facts_cache.json` — list of typed Fact records, staged for Neo4j write.

## Setup & imports

````python
import json
import re
import sys
import hashlib
from decimal import Decimal, InvalidOperation
from enum import Enum
from pathlib import Path
from typing import Literal

import dateparser
from babel.numbers import get_currency_precision
from dotenv import load_dotenv
from openai import OpenAI
from price_parser import Price
from pydantic import BaseModel, Field, model_validator

load_dotenv()
client = OpenAI()

ENTITY_CACHE   = Path("entity_cache.json")
RESOLVED_CACHE = Path("resolved_entities.json")
FACTS_CACHE    = Path("facts_cache.json")

EXTRACTOR_VERSION = "fact-extraction-v0/gpt-4.1-mini"

for p in (ENTITY_CACHE, RESOLVED_CACHE):
    if not p.exists():
        sys.exit(f"missing cache: {p}")
print("caches found ✓")
````

## Pydantic contract — Fact record

Single source of truth for the data shape.  
The LLM fills semantic slots and `raw_value_str`. Typed scalar fields are filled by the normalizer (Stage 2), never by the LLM.

````python
class Predicate(str, Enum):
    uses_tool       = "usesTool"
    owner_of        = "ownerOf"
    has_role        = "hasRole"
    works_for       = "worksFor"
    documented_in   = "documentedIn"
    headcount       = "headcount"
    responsible_for = "responsibleFor"
    part_of         = "partOf"
    founding_date   = "foundingDate"
    revenue         = "revenue"
    funding_raised  = "fundingRaised"
    funding_stage   = "fundingStage"
    industry        = "industry"
    changed_on      = "changedOn"
    other           = "other"


# (family, value_type) contract per predicate
PREDICATE_CONTRACT: dict[str, tuple[str | None, str]] = {
    "usesTool":      ("relational", "entity_ref"),
    "ownerOf":       ("relational", "entity_ref"),
    "hasRole":       ("attribute",  "text"),
    "worksFor":      ("relational", "entity_ref"),
    "documentedIn":  ("relational", "entity_ref"),
    "headcount":     ("attribute",  "count"),
    "responsibleFor":("relational", "entity_ref"),
    "partOf":        ("relational", "entity_ref"),
    "foundingDate":  ("attribute",  "date"),
    "revenue":       ("attribute",  "money"),
    "fundingRaised": ("attribute",  "money"),
    "fundingStage":  ("attribute",  "text"),
    "industry":      ("attribute",  "text"),
    "changedOn":     ("attribute",  "date"),
    "other":         (None,         "text"),
}


class FactRecord(BaseModel):
    """Reified Fact node contract. Filled across pipeline stages."""

    # identity
    fact_id: str = Field(description="stable hash of (doc_id, char_start, predicate, subject)")

    # semantic slots (LLM fills these)
    subject_entity_id: str   = Field(description="canonical resolved entity ID")
    predicate:         Predicate
    predicate_raw:     str   = Field(default="", description="LLM's raw predicate; required when predicate=other")
    family:            Literal["attribute", "relational", "unknown"] = "unknown"
    raw_value_str:     str   = Field(description="verbatim value copied from source")

    # relational value (for entity_ref predicates)
    object_entity_id:  str | None = None

    # typed attribute values (Stage 2 fills at most one block per fact)
    amount_minor:  int | None = None   # money: integer minor units
    currency:      str | None = None   # money: ISO 4217
    count:         int | None = None   # count
    approximate:   bool       = False  # count: was value prefixed with ~
    basis_points:  int | None = None   # percentage
    iso_date:      str | None = None   # date
    iso_duration:  str | None = None   # duration
    text_value:    str | None = None   # text / other

    # bi-temporal validity
    valid_from:        str | None = None
    valid_to:          str | None = None
    temporal_inferred: bool       = False

    # provenance
    doc_id:       str
    char_start:   int
    char_end:     int
    source_quote: str = Field(description="verbatim slice of source text")

    # pipeline stamps
    span_verified:      bool  = False
    norm_ok:            bool  = False
    confidence:         float = 1.0
    extractor_version:  str   = EXTRACTOR_VERSION


def make_fact_id(doc_id: str, char_start: int, predicate: str, subject: str) -> str:
    key = f"{doc_id}|{char_start}|{predicate}|{subject}"
    return hashlib.sha1(key.encode()).hexdigest()[:16]


print("Pydantic contract loaded ✓")
print(f"  {len(Predicate)} predicates  ({sum(1 for f,_ in PREDICATE_CONTRACT.values() if f=='relational')} relational, "
      f"{sum(1 for f,_ in PREDICATE_CONTRACT.values() if f=='attribute')} attribute)")
````

## Load caches

Builds two lookup structures the pipeline needs at every stage:
- `mention_to_canonical` — surface form → canonical entity ID
- `chunks` — deduplicated (episode, source_file) pairs with recovered char offsets

````python
# ── resolved entities ─────────────────────────────────────────────────────────
resolved_data = json.loads(RESOLVED_CACHE.read_text())
mention_to_canonical: dict[str, str] = resolved_data["mention_to_canonical"]
print(f"resolved entities : {resolved_data['total_entities']:>5}")
print(f"surface forms     : {len(mention_to_canonical):>5}")

# ── entity cache — deduplicate chunks ────────────────────────────────────────
entity_cache = json.loads(ENTITY_CACHE.read_text())
mentions     = entity_cache["mentions"]

# One chunk = (source_file, episode_idx). Recover char offsets by searching
# the episode text in the source document. Cache doc text to avoid re-reading.
_doc_cache: dict[str, str] = {}

def _doc_text(source_file: str) -> str:
    if source_file not in _doc_cache:
        p = Path(source_file)
        _doc_cache[source_file] = p.read_text(encoding="utf-8", errors="ignore") if p.exists() else ""
    return _doc_cache[source_file]


# Deduplicate: one chunk entry per (source_file, episode_idx)
ChunkKey = tuple[str, str]  # (source_file, episode_idx)
chunks: dict[ChunkKey, dict] = {}

for m in mentions:
    key: ChunkKey = (m["source_file"], str(m["episode_idx"]))
    if key in chunks:
        continue
    episode = m["episode"]
    doc     = _doc_text(m["source_file"])
    offset  = doc.find(episode)
    chunks[key] = {
        "doc_id":     m["source_file"],
        "episode_idx": str(m["episode_idx"]),
        "text":       episode,
        "char_start": offset,
        "char_end":   offset + len(episode) if offset >= 0 else -1,
    }

# Collect known entities per chunk (surface forms that appear in the episode)
chunk_entities: dict[ChunkKey, list[str]] = {k: [] for k in chunks}
for m in mentions:
    key = (m["source_file"], str(m["episode_idx"]))
    canon = mention_to_canonical.get(m["text"])
    if canon and canon not in chunk_entities[key]:
        chunk_entities[key].append(canon)

bad_offsets = sum(1 for c in chunks.values() if c["char_start"] < 0)
print(f"\nchunks            : {len(chunks):>5}")
print(f"bad char offsets  : {bad_offsets:>5}  (source file missing/moved)")
````

## Stage 1 — Schema-guided LLM extraction

For each chunk the model receives:
- the episode text
- the list of **known canonical entity IDs** present in that chunk

It emits raw extraction records: subject entity, predicate (from the closed enum), and `raw_value_str` copied **verbatim** from the source.  
It must NOT compute or normalise the value — that is Stage 2.

````python
PREDICATE_GLOSSES = """\
usesTool       [relational] — subject (person/team/process) uses tool/system X
ownerOf        [relational] — subject is accountable for X (process, tool, budget, domain)
hasRole        [attribute/text] — subject holds role or title X (role stays a string, not an entity)
worksFor       [relational] — person works for org or team X
documentedIn   [relational] — knowledge/process X lives in source Y (doc, wiki, video)
headcount      [attribute/count] — size of org or team in people
responsibleFor [relational] — subject leads or performs project/task X
partOf         [relational] — team/process is part of larger org or department
foundingDate   [attribute/date] — when an org was founded
revenue        [attribute/money] — org revenue
fundingRaised  [attribute/money] — amount an org has raised
fundingStage   [attribute/text] — seed / Series A / etc.
industry       [attribute/text] — org vertical (fintech, biotech, …)
changedOn      [attribute/date] — a stated event or change date
other          — no predicate fits; fill predicate_raw with what you wanted"""

EXTRACT_SYSTEM = f"""You extract typed facts from a text chunk.

You will receive:
- TEXT: the source chunk
- ENTITIES: canonical entity IDs present in this chunk

For each fact you find:
- subject_entity_id: copy VERBATIM from the ENTITIES list. Only use provided entity IDs.
- predicate: choose from the closed enum. Use "other" if none fits; fill predicate_raw.
- raw_value_str: copy the value VERBATIM from the source text, character-for-character.
  For relational facts this is the object entity name as written in the text.
  Do NOT compute, normalise, or paraphrase the value.
- valid_from / valid_to: ISO dates ONLY if stated explicitly in the text. Leave null if implied.
- temporal_inferred: true if you inferred valid-time rather than reading it from the text.

Predicate glosses:
{PREDICATE_GLOSSES}

Extract only facts where subject is a known entity from the ENTITIES list.
Skip facts with no clear subject. When in doubt, use other + predicate_raw."""

EXTRACT_SCHEMA = {
    "type": "object",
    "additionalProperties": False,
    "properties": {
        "facts": {
            "type": "array",
            "items": {
                "type": "object",
                "additionalProperties": False,
                "properties": {
                    "subject_entity_id": {"type": "string"},
                    "predicate":         {"type": "string", "enum": [p.value for p in Predicate]},
                    "predicate_raw":     {"type": "string"},
                    "raw_value_str":     {"type": "string"},
                    "valid_from":        {"type": ["string", "null"]},
                    "valid_to":          {"type": ["string", "null"]},
                    "temporal_inferred": {"type": "boolean"},
                },
                "required": ["subject_entity_id", "predicate", "predicate_raw",
                             "raw_value_str", "valid_from", "valid_to", "temporal_inferred"],
            },
        }
    },
    "required": ["facts"],
}

EXTRACT_MODEL = "gpt-4.1-mini"


def extract_facts_from_chunk(chunk: dict, entities: list[str]) -> list[dict]:
    """Stage 1: LLM extracts raw fact records from one chunk."""
    if not entities:
        return []
    user_msg = (
        f"TEXT:\n{chunk['text']}\n\n"
        f"ENTITIES:\n" + "\n".join(f"- {e}" for e in entities)
    )
    resp = client.chat.completions.create(
        model=EXTRACT_MODEL,
        temperature=0,
        messages=[
            {"role": "system", "content": EXTRACT_SYSTEM},
            {"role": "user",   "content": user_msg},
        ],
        response_format={
            "type": "json_schema",
            "json_schema": {"name": "facts", "strict": True, "schema": EXTRACT_SCHEMA},
        },
    )
    return json.loads(resp.choices[0].message.content)["facts"]


print("Stage 1 ready ✓")
````

## Stage 2 — Deterministic normalizer

Pure Python. Parses `raw_value_str` into a typed scalar.  
Rule: **if parsing fails, leave typed fields null and mark `norm_ok=False`**. Never guess.

Confidence formula (applied in Stage 4):
```
start          1.00
norm_ok=False  −0.25
span_verified=False  −0.40
temporal_inferred    −0.10
predicate=other      −0.10
```

````python
# ── ISO 4217 currency symbol → code map (common subset) ──────────────────────
_SYMBOL_TO_ISO: dict[str, str] = {
    "$": "USD", "€": "EUR", "£": "GBP", "¥": "JPY",
    "₹": "INR", "₩": "KRW", "₪": "ILS", "₺": "TRY",
    "A$": "AUD", "C$": "CAD", "HK$": "HKD", "S$": "SGD",
}

_APPROX_RE = re.compile(r"^[~≈≤≥><circa ]*", re.IGNORECASE)
_COUNT_RE  = re.compile(r"[\d,]+")
_PCT_RE    = re.compile(r"([\d.]+)\s*%")


def _normalize_money(raw: str) -> dict:
    """price-parser → Decimal → integer minor units. Never float."""
    p = Price.fromstring(raw)
    if p.amount is None:
        return {"norm_ok": False}
    symbol   = (p.currency or "").strip()
    iso      = _SYMBOL_TO_ISO.get(symbol, symbol.upper() or "USD")
    try:
        precision   = get_currency_precision(iso)
    except Exception:
        precision   = 2
    amount_minor = int(Decimal(str(p.amount)) * (10 ** precision))
    return {"norm_ok": True, "amount_minor": amount_minor, "currency": iso}


def _normalize_count(raw: str) -> dict:
    approx   = bool(_APPROX_RE.match(raw).group(0).strip())
    digits   = _COUNT_RE.search(raw.replace(",", ""))
    if not digits:
        return {"norm_ok": False}
    return {"norm_ok": True, "count": int(digits.group()), "approximate": approx}


def _normalize_percentage(raw: str) -> dict:
    m = _PCT_RE.search(raw)
    if not m:
        return {"norm_ok": False}
    try:
        bp = int(Decimal(m.group(1)) * 100)
        return {"norm_ok": True, "basis_points": bp}
    except InvalidOperation:
        return {"norm_ok": False}


def _normalize_date(raw: str) -> dict:
    dt = dateparser.parse(raw, settings={"RETURN_AS_TIMEZONE_AWARE": False})
    if dt is None:
        return {"norm_ok": False}
    return {"norm_ok": True, "iso_date": dt.date().isoformat()}


def _normalize_duration(raw: str) -> dict:
    # dateparser doesn't handle ISO durations natively; store raw for now
    # upgrade path: Duckling service or isodate library
    return {"norm_ok": True, "iso_duration": raw}


def normalize(predicate: str, raw_value_str: str) -> dict:
    """Stage 2: dispatch to the right parser based on predicate contract."""
    _, vtype = PREDICATE_CONTRACT.get(predicate, (None, "text"))
    if vtype == "money":      return _normalize_money(raw_value_str)
    if vtype == "count":      return _normalize_count(raw_value_str)
    if vtype == "percentage": return _normalize_percentage(raw_value_str)
    if vtype == "date":       return _normalize_date(raw_value_str)
    if vtype == "duration":   return _normalize_duration(raw_value_str)
    # text / entity_ref / other — no parsing needed
    return {"norm_ok": True, "text_value": raw_value_str}


print("Stage 2 normalizer ready ✓")
````

## Stage 3 — Span re-verification

Deterministically confirms `raw_value_str` (or its digit sequence) appears inside `source_quote`.  
If absent the LLM likely paraphrased or hallucinated — `span_verified=False` drops confidence sharply.

Also checks temporal slots: if `valid_from`/`valid_to` don't appear in the source text, sets `temporal_inferred=True`.

````python
_DIGITS_RE = re.compile(r"\d+")


def verify_span(raw_value_str: str, source_quote: str,
                valid_from: str | None, valid_to: str | None) -> dict:
    """Stage 3: check raw_value_str and temporal slots appear in source_quote."""
    quote_lower = source_quote.lower()
    raw_lower   = raw_value_str.lower().strip()

    # primary check: verbatim substring
    if raw_lower and raw_lower in quote_lower:
        span_verified = True
    else:
        # fallback: all digit sequences from raw appear in quote
        digits = _DIGITS_RE.findall(raw_value_str)
        span_verified = bool(digits) and all(d in source_quote for d in digits)

    # temporal check: if a date was stated, its digits should be in the source
    temporal_inferred = False
    for date_val in filter(None, [valid_from, valid_to]):
        date_digits = _DIGITS_RE.findall(date_val)
        if date_digits and not all(d in source_quote for d in date_digits):
            temporal_inferred = True
            break

    return {"span_verified": span_verified, "temporal_inferred": temporal_inferred}


print("Stage 3 verifier ready ✓")
````

## Stage 4 — Confidence scoring

Rule-based composite. **Not** model logprob-based.

```
start                1.00
norm_ok=False       −0.25   (value couldn't be parsed)
span_verified=False −0.40   (value not found in source — highest risk)
temporal_inferred   −0.10   (valid-time was inferred, not stated)
predicate=other     −0.10   (no predicate fit)
```
Floor: 0.0

````python
def score_confidence(norm_ok: bool, span_verified: bool,
                     temporal_inferred: bool, predicate: str) -> float:
    """Stage 4: rule-based confidence score in [0, 1]."""
    score = 1.00
    if not norm_ok:        score -= 0.25
    if not span_verified:  score -= 0.40
    if temporal_inferred:  score -= 0.10
    if predicate == "other": score -= 0.10
    return round(max(0.0, score), 2)


print("Stage 4 confidence scorer ready ✓")
````

## Full pipeline — assemble one FactRecord per LLM extraction

````python
def process_chunk(chunk: dict, entities: list[str]) -> list[FactRecord]:
    """Run all four stages on one chunk and return FactRecords."""
    raw_facts = extract_facts_from_chunk(chunk, entities)  # Stage 1
    records: list[FactRecord] = []

    for rf in raw_facts:
        pred     = rf["predicate"]
        raw_val  = rf["raw_value_str"]
        family, _ = PREDICATE_CONTRACT.get(pred, ("unknown", "text"))

        source_quote = chunk["text"]
        char_start   = chunk["char_start"]
        char_end     = chunk["char_end"]

        # resolve relational object → canonical ID
        object_entity_id = (
            mention_to_canonical.get(raw_val)
            if family == "relational" else None
        )

        norm   = normalize(pred, raw_val)                          # Stage 2
        verify = verify_span(raw_val, source_quote,               # Stage 3
                             rf.get("valid_from"), rf.get("valid_to"))
        conf   = score_confidence(                                  # Stage 4
            norm_ok          = norm.get("norm_ok", False),
            span_verified    = verify["span_verified"],
            temporal_inferred= verify["temporal_inferred"] or rf.get("temporal_inferred", False),
            predicate        = pred,
        )

        record = FactRecord(
            fact_id          = make_fact_id(chunk["doc_id"], char_start, pred, rf["subject_entity_id"]),
            subject_entity_id= rf["subject_entity_id"],
            predicate        = Predicate(pred),
            predicate_raw    = rf.get("predicate_raw", ""),
            family           = family or "unknown",
            raw_value_str    = raw_val,
            object_entity_id = object_entity_id,
            valid_from       = rf.get("valid_from"),
            valid_to         = rf.get("valid_to"),
            temporal_inferred= verify["temporal_inferred"] or rf.get("temporal_inferred", False),
            doc_id           = chunk["doc_id"],
            char_start       = char_start,
            char_end         = char_end,
            source_quote     = source_quote,
            span_verified    = verify["span_verified"],
            norm_ok          = norm.get("norm_ok", False),
            confidence       = conf,
            **{k: v for k, v in norm.items() if k != "norm_ok"},
        )
        records.append(record)

    return records


print("pipeline assembled ✓")
````

## Demo run — 5 chunks

Picks the first 5 chunks that have at least one known entity. Prints typed Fact records with confidence and verification flags.  
Do **not** run the full corpus here — use the bulk cell below.

````python
DEMO_CHUNKS = 5

demo_records: list[FactRecord] = []

for key, chunk in list(chunks.items()):
    if len(demo_records) >= DEMO_CHUNKS * 3:  # rough cap
        break
    entities = chunk_entities.get(key, [])
    if not entities or chunk["char_start"] < 0:
        continue
    try:
        demo_records.extend(process_chunk(chunk, entities))
    except Exception as e:
        print(f"  chunk {key} failed: {e}")
    if len(demo_records) // 3 >= DEMO_CHUNKS:
        break

print(f"\n{len(demo_records)} facts extracted from demo chunks\n")

for r in demo_records:
    ok  = "✓" if r.span_verified else "✗"
    nrm = "✓" if r.norm_ok      else "✗"
    tin = " ⏱inferred" if r.temporal_inferred else ""
    print(f"[{r.confidence:.2f}] span={ok} norm={nrm}{tin}")
    print(f"  subject : {r.subject_entity_id}")
    print(f"  predicate: {r.predicate.value}" + (f" ({r.predicate_raw})" if r.predicate_raw else ""))
    print(f"  raw     : {r.raw_value_str!r}")

    # show typed value
    if r.amount_minor is not None:
        print(f"  typed   : {r.amount_minor} {r.currency} minor-units")
    elif r.count is not None:
        approx = "~" if r.approximate else ""
        print(f"  typed   : {approx}{r.count} (count)")
    elif r.basis_points is not None:
        print(f"  typed   : {r.basis_points} bp")
    elif r.iso_date:
        print(f"  typed   : {r.iso_date} (date)")
    elif r.text_value:
        print(f"  typed   : {r.text_value!r} (text)")
    elif r.object_entity_id:
        print(f"  object  : {r.object_entity_id}")

    print(f"  source  : {Path(r.doc_id).name}  [{r.char_start}:{r.char_end}]")
    print()
````

## Bulk run — all chunks → `facts_cache.json`

Processes every chunk that has at least one known entity and a valid char offset.  
Results are saved incrementally every 50 chunks.

````python
from concurrent.futures import ThreadPoolExecutor, as_completed
from tqdm.notebook import tqdm as tqdm_nb

BULK_WORKERS   = 8
SAVE_EVERY     = 50

eligible = [
    (key, chunk)
    for key, chunk in chunks.items()
    if chunk["char_start"] >= 0 and chunk_entities.get(key)
]
print(f"{len(eligible)} eligible chunks (have entities + valid offset)")

all_facts: list[dict] = []

def _process(args: tuple) -> list[dict]:
    key, chunk = args
    records = process_chunk(chunk, chunk_entities[key])
    return [r.model_dump(mode="json") for r in records]


def _save() -> None:
    FACTS_CACHE.write_text(json.dumps(all_facts, ensure_ascii=False, indent=2))


with ThreadPoolExecutor(max_workers=BULK_WORKERS) as executor:
    futures = {executor.submit(_process, item): item for item in eligible}
    with tqdm_nb(total=len(eligible), desc="extracting facts") as bar:
        for i, future in enumerate(as_completed(futures)):
            all_facts.extend(future.result())
            bar.update(1)
            if (i + 1) % SAVE_EVERY == 0:
                _save()

_save()

conf_vals = [f["confidence"] for f in all_facts]
verified  = sum(1 for f in all_facts if f["span_verified"])
norm_ok   = sum(1 for f in all_facts if f["norm_ok"])
by_pred   = {}
for f in all_facts:
    by_pred[f["predicate"]] = by_pred.get(f["predicate"], 0) + 1

print(f"\n{len(all_facts)} facts saved → {FACTS_CACHE}")
print(f"  span verified : {verified}/{len(all_facts)} ({100*verified//max(1,len(all_facts))}%)")
print(f"  norm ok       : {norm_ok}/{len(all_facts)} ({100*norm_ok//max(1,len(all_facts))}%)")
print(f"  avg confidence: {sum(conf_vals)/max(1,len(conf_vals)):.2f}")
print("\n  by predicate:")
for pred, cnt in sorted(by_pred.items(), key=lambda x: -x[1]):
    print(f"    {pred:<20} {cnt}")
````

1. `NEO4J_PASSWORD` in `.env`
2. A Fact node schema / constraints defined in the database
